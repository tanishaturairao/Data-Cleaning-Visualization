```python

 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")

# Create the correct output folder
output_folder = "C:/Users/faizan/OneDrive/Desktop/Thiranex-Data-Science/visualizations"
os.makedirs(output_folder, exist_ok=True)

# Load cleaned datasets
orders = pd.read_csv(
    "C:/Users/faizan/OneDrive/Desktop/Thiranex-Data-Science/data/processed/orders_cleaned.csv",
    parse_dates=["order_purchase_timestamp"]
)
customers = pd.read_csv("C:/Users/faizan/OneDrive/Desktop/Thiranex-Data-Science/data/processed/customers_cleaned.csv")
payments = pd.read_csv("C:/Users/faizan/OneDrive/Desktop/Thiranex-Data-Science/data/processed/payments_cleaned.csv")

# 1. Order status distribution
plt.figure(figsize=(9, 5))
sns.countplot(data=orders, x="order_status")
plt.title("Distribution of Order Status")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(output_folder, "order_status.png"))
plt.show()

# 2. Monthly order trends
orders["month"] = orders["order_purchase_timestamp"].dt.to_period("M").astype(str)
monthly_orders = orders.groupby("month").size()

plt.figure(figsize=(12, 5))
monthly_orders.plot(kind="line", marker="o")
plt.title("Monthly Order Trends")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(output_folder, "monthly_orders.png"))
plt.show()

# 3. Payment type distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=payments, x="payment_type")
plt.title("Distribution of Payment Types")
plt.xlabel("Payment Type")
plt.ylabel("Number of Payments")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(os.path.join(output_folder, "payment_types.png"))
plt.show()

# 4. Top 10 customer states
top_states = customers["customer_state"].value_counts().head(10)

plt.figure(figsize=(9, 5))
top_states.plot(kind="bar")
plt.title("Top 10 States by Customer Records")
plt.xlabel("State")
plt.ylabel("Number of Customer Records")
plt.tight_layout()
plt.savefig(os.path.join(output_folder, "top_customer_states.png"))
plt.show()

print("VISUALIZATION COMPLETED!")
print("Charts saved in:", os.path.abspath(output_folder))
```
