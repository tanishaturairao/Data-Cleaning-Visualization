```python

# Define dataset path

orders_path = "C:/Users/faizan/OneDrive/Desktop/Thiranex-Data-Science/data/raw/olist_customers_dataset.csv"

# Load dataset
orders = pd.read_csv(orders_path)

# Display first five records
orders.head()
```

```python

# 1. Check the number of rows and columns
print("Dataset Shape:")
print(orders.shape)

# 2. Display column names
print("\nColumn Names:")
print(orders.columns.tolist())

# 3. Understand the dataset
print("\nDataset Information:")
orders.info()
```

```python

# Check missing values in every column

missing_values = orders.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python

# Calculate missing value percentages

missing_percentage = (
    orders.isnull().sum() / len(orders)
) * 100

print("Missing Values Percentage:")
print(missing_percentage.round(2))
```

```python

 
# Count missing values
missing_values = orders.isnull().sum()

# Keep columns containing missing values
missing_values = missing_values[missing_values > 0]

# Check whether there is anything to visualize
if len(missing_values) > 0:

    plt.figure(figsize=(10, 5))
    missing_values.plot(kind="bar")

    plt.title("Missing Values by Column")
    plt.xlabel("Columns")
    plt.ylabel("Number of Missing Values")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

else:
    print("Great! No missing values found in this dataset.")
```

```python
# Create a separate working copy
customers = customers.copy()

# Convert ZIP code prefix to string
customers["customer_zip_code_prefix"] = (
    customers["customer_zip_code_prefix"].astype(str)
)

# Remove unnecessary spaces from text columns
text_columns = ["customer_city", "customer_state"]

for col in text_columns:
    customers[col] = customers[col].str.strip()

# Standardize text capitalization
customers["customer_city"] = customers["customer_city"].str.lower()
customers["customer_state"] = customers["customer_state"].str.upper()

# Verify the cleaned dataset
print("Dataset shape:", customers.shape)
print("\nMissing values:")
print(customers.isnull().sum())
print("\nData types:")
print(customers.dtypes)
print("\nPreview:")
display(customers.head())
```
